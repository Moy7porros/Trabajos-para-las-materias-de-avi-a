# Parte 1: ¿Cómo funciona una red neuronal?

## Introducción

Las redes neuronales son estructuras inspiradas en el funcionamiento del cerebro humano y son ampliamente utilizadas dentro del área de Machine Learning debido a su capacidad para aprender patrones complejos y realizar predicciones.

Estas redes están compuestas por unidades llamadas **neuronas artificiales** o **perceptrones**, las cuales trabajan en conjunto para procesar información y tomar decisiones. Gracias a la gran cantidad de parámetros ajustables que poseen, las redes neuronales pueden resolver problemas de clasificación, regresión, reconocimiento de voz, procesamiento de imágenes y generación de texto, entre muchas otras aplicaciones.

---

## El perceptrón: la neurona básica

El elemento más simple dentro de una red neuronal es el **perceptrón**.  
Este recibe diferentes entradas o características y genera una salida dependiendo de los valores recibidos.

Por ejemplo, para decidir si una persona irá a un concierto se podrían utilizar las siguientes entradas:

- Clima agradable
- Tener dinero
- Que los amigos asistan

Cada una de estas entradas puede representarse con valores numéricos:

- `1` = Sí
- `0` = No

Estas entradas son conocidas como variables o características de entrada.

---

## Pesos y entradas ponderadas

No todas las entradas tienen la misma importancia.  
En una red neuronal cada entrada posee un valor llamado **peso** (*weight*).

Los pesos permiten que ciertas entradas influyan más que otras en la decisión final.

Por ejemplo:

- El clima podría ser más importante que tener dinero.
- La asistencia de amigos podría tener menor importancia.

La red neuronal ajusta estos pesos durante el entrenamiento para mejorar sus predicciones.

---

## Umbral o sesgo (Bias)

Además de los pesos, el perceptrón cuenta con un valor adicional llamado **umbral** o **sesgo (bias)**.

El sesgo ayuda a determinar si la suma de las entradas ponderadas es suficiente para activar la neurona y producir una salida.

El proceso básico es:

1. Se multiplican las entradas por sus pesos.
2. Se suman todos los resultados.
3. Se agrega el sesgo.
4. Se genera una salida.

De esta forma, el perceptrón puede tomar decisiones simples.

---

## Redes neuronales multicapa

Un único perceptrón puede resolver problemas sencillos, pero para problemas más complejos se conectan múltiples perceptrones formando una **red neuronal multicapa**.

En este tipo de red:

- La salida de una neurona sirve como entrada para otras neuronas.
- Existen múltiples capas de procesamiento.
- Cada capa aprende diferentes patrones de los datos.

Gracias a esto, las redes neuronales pueden encontrar relaciones complejas dentro de la información.

---

## Aprendizaje de la red neuronal

Las redes neuronales aprenden ajustando automáticamente sus pesos y sesgos.

Durante el entrenamiento:

1. La red realiza una predicción.
2. Se compara la predicción con el valor real.
3. Se calcula el error.
4. Los pesos se modifican para reducir ese error.

Este proceso se repite muchas veces hasta que la red logra generar predicciones precisas.

---

## Aplicaciones de las redes neuronales

Debido a su capacidad de aprendizaje, las redes neuronales pueden utilizarse en distintos problemas como:

- Regresión
- Clasificación
- Reconocimiento de voz
- Procesamiento de imágenes
- Generación de texto
- Traducción automática
- Sistemas inteligentes

Actualmente son una de las tecnologías más importantes dentro de la Inteligencia Artificial y el Deep Learning.

---

## Conclusión

Las redes neuronales funcionan mediante la conexión de múltiples perceptrones capaces de aprender patrones a partir de datos.  
Cada neurona utiliza entradas, pesos y sesgos para generar decisiones, mientras que el entrenamiento permite ajustar automáticamente los parámetros para mejorar el rendimiento del modelo.

Gracias a esta capacidad de aprendizaje, las redes neuronales pueden resolver problemas complejos y son utilizadas ampliamente en aplicaciones modernas de Inteligencia Artificial.

# Parte 2: Red neuronal de regresión

## Introducción

En esta práctica se implementó una red neuronal utilizando TensorFlow para resolver un problema de regresión lineal.

El objetivo del modelo es aprender la relación entre grados Celsius y grados Fahrenheit utilizando una red neuronal simple.

Para este ejercicio se utilizaron datos previamente preparados y se entrenó el modelo directamente sin separar datos de entrenamiento y prueba, debido a que se trata de un problema sencillo de conversión matemática.

---

## Librerías utilizadas

Para desarrollar el modelo se utilizaron las siguientes bibliotecas:

- **Pandas**: para cargar y manipular los datos.
- **TensorFlow**: para crear y entrenar la red neuronal.
- **Numpy**: para hacer arrays de numpy
---

## Carga de datos

Primero se importó el archivo que contiene los datos de temperatura en Celsius y Fahrenheit.


```python
import pandas as pd
datos = pd.read_csv("celsius.csv")
```

**Visualizacion de datos inicial**

La función head() permite visualizar los primeros registros del conjunto de datos.

```python
datos.head()
```

## Extracción de Características (Variables de Entrada)

En esta línea se extraen los datos de la columna `celsius` para prepararlos como el conjunto de características ($X$) que se alimentará al modelo de aprendizaje automático.



```python
datos["celsius"].values
```

##Definición de variables

Posteriormente se definieron las variables de entrada (x) y salida (y).

* x representa los grados Celsius.

* y representa los grados Fahrenheit.

La propiedad .values convierte las columnas de Pandas en arreglos numéricos que pueden ser utilizados por TensorFlow.

```python
x= datos["celsius"].values
y= datos["fahrenheit"].values
```

##Importación de TensorFlow

Para crear la red neuronal se utilizó TensorFlow.

TensorFlow es una de las bibliotecas más utilizadas para Deep Learning y redes neuronales.

##Creación de la red neuronal

La red neuronal fue creada utilizando una capa densa con una neurona.

En este caso:

* Dense crea una capa completamente conectada.
* units=1 indica que existe una sola neurona.
* input_shape=[1] indica que el modelo recibe un único valor de entrada.

##Compilación del modelo

Después se configuró el modelo indicando el optimizador y la función de pérdida.

**Explicación**

* **Adam:** optimizador encargado de ajustar automáticamente los pesos de la red neuronal.
* **Mean Squared Error (MSE):** mide qué tan grande es el error entre las predicciones y los valores reales.

```python
import tensorflow as tf

modelo = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(1,)), # La entrada son los celsius
    tf.keras.layers.Dense(units=1), # La salida tambien es un dato
])

modelo.compile(
    optimizer=tf.keras.optimizers.Adam(.1),
    loss="mean_squared_error"
)
```

##Entrenamiento del modelo

Una vez configurado el modelo, se realizó el entrenamiento.

**Parámetros utilizados**
* **x:** datos de entrada.
* **y:** valores esperados.
* **epochs=1000:** número de iteraciones de entrenamiento.
* **verbose=False:** evita mostrar demasiada información durante el entrenamiento.

Durante este proceso la red neuronal ajusta automáticamente sus pesos y sesgos para reducir el error.

```python
Entrenamiento  = modelo.fit(x, y, epochs=1000)
```

##Predicción

Finalmente se utilizó el modelo entrenado para realizar predicciones.

El modelo recibe un valor en grados Celsius y devuelve su equivalente aproximado en Fahrenheit.

```python
import matplotlib.pyplot as plt
plt.xlabel("# Epoca")
plt.ylabel("Magnitud de perdida")
plt.plot(Entrenamiento.history["loss"])
```

#Funcionamiento del aprendizaje

La red neuronal aprende encontrando patrones matemáticos entre los datos de entrada y salida.

En este caso, aprende la relación entre Celsius y Fahrenheit:

F = (C × 1.8) + 32

Aunque la fórmula no se programa manualmente, la red neuronal logra aproximarla ajustando sus pesos durante el entrenamiento.

```python
import numpy as np

c = 1230

# Convertimos la lista a un array de NumPy
entrada = np.array([[c]], dtype=float)

# Hacemos la predicción
prediccion = modelo.predict(entrada)

# Ojo: prediccion[0][0] para acceder al valor numérico exacto dentro de la matriz de salida
print(f"Predicción: {c} celsius son {prediccion[0][0]:.2f} fahrenheit")
```
